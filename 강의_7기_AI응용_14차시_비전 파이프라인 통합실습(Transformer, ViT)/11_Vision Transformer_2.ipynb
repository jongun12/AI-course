* 트랜스포머는 2017년 구글이 발표한 "Attention is all you need"에서 나온 자연어 처리 (Natural Language Processing, NLP)모델
* 순환 신경망 모델 (Recurrent Neural Network, RNN)을 사용하지 않고 인코더-디코더 구조로만 구성했음에도 번역 성능에서 RNN보다 우수한 성능을 보임
* 후에 새로운 자연어 처리 모델뿐만 아니라 비전 분야에도 큰 영향을 끼쳐서 트랜스포머 기반 비전 모델이 많이 탄생함

## Positional Encoding

* 트랜스포머는 단어 입력을 순차로 받는 방식이 아니므로 단어의 위치 정보를 알려줘야 함
* 각 단어의 임베딩 벡터에 위치 정보를 더하여 모델의 입력으로 사용하는 것을 포지셔널 인코딩(Positional Encoding)이라 함

## Attention

* Attention mechanism: 디코더 매 시점마다 인코더의 모든 입력 단어들의 hidden state를 참고

* **Self-Attention**: 입력 단어들끼리 위치 정보를 표현 및 연관성을 학습

Query, Key, Value

* Query: "현재 보고 있는 단어가 다른 단어들로부터 어떤 정보를 필요로 하는가?"를 나타내는 벡터
* Key: "이 단어는 어떤 질문(query)에 답할 수 있는가?"를 나타내는 벡터. Query와 내적했을 때 그 값이 크면 유사도가 큰 것임. 즉, 주목(attention)을 많이 받는다는 뜻. Query와 Key랑 내적한 값을 Attention 스코어라 한다.
* Value: 실제로 전달해줄 내용을 담는 벡터. Attention 스코어에 따라 합산되어 최종 문맥 벡터(context vector)를 만든다. 즉, 각 단어가 가지는 정보를 뜻함.

## ViT Code

```python
import torch
import torch.nn as nn
from einops.layers.torch import Rearrange

# ViT 인코더
class Encoder(nn.Module):
    def __init__(self, embed_size=768, num_heads=3, dropout=0.1):
        super().__init__()
        # Multi-head attention layer 추가
        self.attention = nn.MultiheadAttention(embed_size, num_heads, dropout, batch_first=True)
        # Layer Normalization 1 추가. 출력 dimension은 embed_size
        self.layernorm1 = nn.LayerNorm(embed_size)
        # Layer Normalization 2 추가. 출력 dimension은 embed_size
        self.layernorm2 = nn.LayerNorm(embed_size) 
        # Feedforward network 추가 (Linear->GELU->dropout->Linear->dropout)
        self.feedfoward = nn.Sequential(
            nn.Linear(embed_size, 4 * embed_size),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(4 * embed_size, embed_size),
            nn.Dropout(dropout)
        )
        # Dropout layer
        self.dropout = nn.Dropout(dropout)

    def forward(self, x):
        # To-do: 정방향 연산 구현
        # layernorm1 -> multi-head attention + 잔차 연결 -> layernorm2 -> feedforward network + 잔차 연결
        return x
    

# ViT 모델 클래스
class VisionTransformer(nn.Module):
    def __init__(self, in_channels=3, num_encoders=6, embed_size=768, img_size=(224, 224), patch_size=16, num_classes=10, num_heads=12):
        super().__init__()
        self.img_size = img_size # 이미지 크기
        self.patch_size = patch_size # 패치 크기
        patch_nums = (img_size[0]*img_size[1])//(patch_size**2) # 이미지 split 했을 때 나오는 토큰 개수

        # CLS_token 선언: nn.Parameter 이용
        self.class_token = nn.Parameter(torch.randn((embed_size,)), requires_grad=True)
        
        # patch 임베딩 (성능 높이기 위해 conv2d 사용)
        self.patch_embedding = nn.Sequential(
            nn.Conv2d(in_channels, embed_size, kernel_size=patch_size, stride=patch_size),
            Rearrange('b e h w -> b (h w) e')
        )
        
        # Position embedding 선언: nn.Parameter 이용
        self.pos_embedding = nn.Parameter(torch.randn((patch_nums+1, embed_size)), requires_grad=True)
        
        # Encoder 클래스를 활용하여 여러겹의 인코더를 쌓은 하나의 self 변수 선언 (nn.ModuleList 활용)
        self.encoders = nn.ModuleList([
            Encoder(embed_size=embed_size, num_heads=num_heads) for _ in range(num_encoders)
        ])
        
        # 분류를 위한 최종 MLP head 구성 (nn.Sequential를 활용하여 Layernorm -> Linear 연산하도록 구성)
        self.mlp_head = nn.Sequential(
            nn.LayerNorm(embed_size),
            nn.Linear(embed_size, num_classes)
        )
        
    def forward(self, x): # 입력 shape: [batch_size, channel, img_size, img_size]
        batch_size = x.shape[0]
        x = self.patch_embedding(x) # [batch_size, patch_nums, emb_size]
        class_token = self.class_token.unsqueeze(0).repeat(batch_size, 1, 1) # class token 추가
        x = torch.cat([class_token, x], dim=1)
        x = x + self.pos_embedding.unsqueeze(0) # positional encoding
        for encoder in self.encoders: # 인코더 통과
            x = encoder(x)
        x = x[:,0,:] # 마지막 인코더에서 나온 cls token
        x = self.mlp_head(x) # cls token을 MLP에 통과하여 최종 output 출력
        return x
```

```python
model = VisionTransformer(num_encoders=6, embed_size=768, img_size=(224, 224), patch_size=16, num_heads=12)
input_ = torch.randn((8,3,224,224)) # 무작위 텐서 입력 shape: [8,3,224,224]
output = model(input_)
```
