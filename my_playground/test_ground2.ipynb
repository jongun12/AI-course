```python
import cv2
from PIL import Image
import os
import numpy as np
import pyheif
import matplotlib.pyplot as plt
```

```python
def target_points(ra=(1.414,1),target=1732):
    """
    ra는 (h,w) 비율
    대각선 길이가 1732를 목표로 함
    """
    b = target / (ra[0]**2+ra[1]**2)**0.5
    ra = np.array(ra)
    hw = ra*b
    return [[0,hw[0].item()],[0,0],[hw[1].item(),0],[hw[1].item(),hw[0].item()]]
```
