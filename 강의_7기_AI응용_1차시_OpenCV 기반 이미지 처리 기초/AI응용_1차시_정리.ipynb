```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
path = '/home/kim/Desktop/AI_cours/data/puppy.jpg'
```

```python
image = cv2.imread(path)
cv2.blur(image, (3,3))
cv2.GaussianBlur(image, (11,11))

gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
binary_image = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY) # cv2.THRESH_OTSU # THRESH_OTSU 자동으로 threshold 값 정함
cv2.adaptiveThreshold

kernel = np.ones((11,11), dtype=np.uint8)
# kernel 사이즈가 아닌 kernel을 받는 이유는 다양한 형태의 커널 사용을 가능하게 하기 위힘이다.
# iterations은 몇 번 반복할 건지
erode_image = cv2.erode(binary_image, kernel, iterations=1) # 침식 (커널 내에 하나의 검은 색이라도 있으면 중앙 픽셀을 0으로)
dilate_image = cv2.dilate(binary_image, kernel, iterations=1) # 팽창 (커널 내에 하나의 흰 색이라도 있으면 중앙 픽셀을 255로)

opening = cv2.morphologyEx(binary_image, cv2.MORPH_OPEN, kernel) # 열기 (침식 -> 팽창) 흰색 노이즈 제거
closing = cv2.morphologyEx(binary_image, cv2.MORPH_CLOSE, kernel) # 닫기 (팽창 -> 침식) 검은색 노이즈 제거
```

```python
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

lower_blue = np.array([90,50,50])
high_blue = np.array([130,255,255])

mask = cv2.inRange(image_hsv, lower_blue, high_blue) # 특정 범위 내에 있으면 255, 아니면 0

# bitwise_and(원본이미지, 원본이미지, mask=마스크이미지)
result = cv2.bitwise_and(image, image, mask=mask)
# cv2.bitwise_and(image, mask)는 채널이 안 맞기 때문에 오류
# mask를 3채널로 늘린 다음 (cv2.cvtColor(mask, cv2.COLOR_GRAY2BGR)
# 연산을 하면 가능하지만
# 일반적으로 위의 코드로 많이 사용됨
```
