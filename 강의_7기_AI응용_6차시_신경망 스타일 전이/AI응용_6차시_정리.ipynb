```python
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
```

```python
# 스타일 전의
# 백본은 이미 훈련된 cnn모델을 사용
# 스타일 전의에서는 모델을 훈련시키는 것이 아니라 노이즈 이미지를 업데이트 시키는 것임!!

# GramMatrix >> 위치 정보는 없애고 스타일만 남김
# 콘텐츠 이미지, 스타일 이미지
# 스타일 이미지는 grammatrix(featuremap)를 사용하여 비교
# 먼저 스타일의 각 지점의 grammatrix(featuremap) 계산
# 콘텐츠 이미지의 한 부분의 피쳐맵 계산
# MSELoss 사용

# 재밌긴 한데, 쓸일은 많이 없을 지도..

# 자세한 부분은 강의 코드 확인

# 위의 두 loss만 하면 점들이 박혀있게 되는데 아래 loss도 같이 사용하면 줄어듬(부드러운 정도를 loss로 사용)
def total_variation_loss(img):
    # Total variation loss란 인접한 픽셀간 차이값이 너무 크지 않도록 규제 
    # 한 픽셀씩 어긋난 픽셀 리스트끼리 빼면 >> 
    loss = torch.sum(torch.abs(img[:,:,:,:-1] - img[:,:,:,1:])) +\
    torch.sum(torch.abs(img[:,:,:-1,:] - img[:,:,1:,:]))
    return loss
```
